## 5. Switch Statements and Control Flow

Switch statements provide an elegant way to handle multiple conditions.

````python
// Switch Statements - Day of Week Example
Console.WriteLine("=== Switch Statements (Day of Week) ===");

// Test different days
for (int day = 1; day <= 8; day++)
{
    Console.Write($"Day {day}: ");
    
    switch (day)
    {
        case 1:
        case 2:
        case 3:
        case 4:
        case 5:
            Console.WriteLine("Work day");
            break;
        case 6:
        case 7:
            Console.WriteLine("Weekend");
            break;
        default:
            Console.WriteLine("Invalid day");
            break;
    }
}

// Modern switch expression (C# 8.0+)
Console.WriteLine("\n=== Modern Switch Expressions ===");

string GetDayType(int dayNumber) => dayNumber switch
{
    1 or 2 or 3 or 4 or 5 => "Workday",
    6 or 7 => "Weekend",
    _ => "Invalid day"
};

for (int day = 1; day <= 8; day++)
{
    Console.WriteLine($"Day {day}: {GetDayType(day)}");
}

// Switch with different data types
Console.WriteLine("\n=== Switch with Different Types ===");

// Switch with characters
char grade = 'B';
string result = grade switch
{
    'A' => "Excellent",
    'B' => "Good",
    'C' => "Average",
    'D' => "Below Average",
    'F' => "Fail",
    _ => "Invalid Grade"
};

Console.WriteLine($"Grade {grade}: {result}");

// Switch with strings
string season = "Summer";
string activity = season switch
{
    "Spring" => "Planting flowers",
    "Summer" => "Going to the beach",
    "Fall" or "Autumn" => "Collecting leaves",
    "Winter" => "Building snowmen",
    _ => "Unknown season activity"
};

Console.WriteLine($"{season}: {activity}");
````

## 6. Loop Examples (For and While)

Loops are fundamental for repeating operations and iterating through data.

````python
// Loop Examples
Console.WriteLine("=== For Loop Examples ===");

// Basic for loop
Console.WriteLine("Counting from 1 to 5:");
for (int i = 1; i <= 5; i++)
{
    Console.WriteLine($"Count: {i}");
}

// For loop with step
Console.WriteLine("\nEven numbers from 2 to 10:");
for (int i = 2; i <= 10; i += 2)
{
    Console.WriteLine($"Even: {i}");
}

// Reverse counting
Console.WriteLine("\nCountdown from 5 to 1:");
for (int i = 5; i >= 1; i--)
{
    Console.WriteLine($"Countdown: {i}");
}

// Nested for loops
Console.WriteLine("\n=== Nested For Loops ===");
Console.WriteLine("Multiplication table snippet:");
for (int i = 1; i <= 3; i++)
{
    for (int j = 1; j <= 3; j++)
    {
        Console.Write($"{i}×{j}={i*j}\t");
    }
    Console.WriteLine();
}

// While loop examples
Console.WriteLine("\n=== While Loop Examples ===");

int counter = 1;
Console.WriteLine("While loop counting to 5:");
while (counter <= 5)
{
    Console.WriteLine($"While count: {counter}");
    counter++;
}

// Do-while loop
Console.WriteLine("\n=== Do-While Loop Example ===");
int number = 10;
do
{
    Console.WriteLine($"Do-while: {number}");
    number--;
} while (number > 7);

// Break and continue examples
Console.WriteLine("\n=== Break and Continue Examples ===");

Console.WriteLine("Numbers 1-10, skipping 5, stopping at 8:");
for (int i = 1; i <= 10; i++)
{
    if (i == 5)
    {
        Console.WriteLine("Skipping 5...");
        continue;
    }
    if (i == 8)
    {
        Console.WriteLine("Breaking at 8!");
        break;
    }
    Console.WriteLine($"Processing: {i}");
}

// Foreach loop with array
Console.WriteLine("\n=== Foreach Loop Example ===");
string[] fruits = { "Apple", "Banana", "Cherry", "Date" };
Console.WriteLine("Fruits list:");
foreach (string fruit in fruits)
{
    Console.WriteLine($"- {fruit}");
}
````

## 7. String Manipulation Examples

Strings are fundamental in C# programming. Let's explore various string operations.

````python
// String Manipulation Examples
Console.WriteLine("=== String Manipulation Examples ===");

string name = "John Doe";
Console.WriteLine($"Original string: '{name}'");

// Basic string properties and methods
Console.WriteLine($"Length: {name.Length}");
Console.WriteLine($"Uppercase: {name.ToUpper()}");
Console.WriteLine($"Lowercase: {name.ToLower()}");

// String indexing and searching
Console.WriteLine($"\n=== String Indexing and Searching ===");
char firstChar = name[0];
char lastChar = name[name.Length - 1];
Console.WriteLine($"First character: '{firstChar}'");
Console.WriteLine($"Last character: '{lastChar}'");

int spaceIndex = name.IndexOf(' ');
Console.WriteLine($"Index of space: {spaceIndex}");

// Substring operations
Console.WriteLine($"\n=== Substring Operations ===");
string firstName = name.Substring(0, spaceIndex);
string lastName = name.Substring(spaceIndex + 1);
Console.WriteLine($"First name: '{firstName}'");
Console.WriteLine($"Last name: '{lastName}'");

// String contains and comparison
Console.WriteLine($"\n=== String Comparison and Contains ===");
Console.WriteLine($"Contains 'John': {name.Contains("John")}");
Console.WriteLine($"Contains 'jane': {name.Contains("jane")}");
Console.WriteLine($"Starts with 'John': {name.StartsWith("John")}");
Console.WriteLine($"Ends with 'Doe': {name.EndsWith("Doe")}");

// String formatting
Console.WriteLine($"\n=== String Formatting ===");
string template = "Hello, {0}! You are {1} years old.";
string formatted = string.Format(template, "Alice", 25);
Console.WriteLine($"String.Format: {formatted}");

// String interpolation
string interpolated = $"Hello, {"Bob"}! You are {30} years old.";
Console.WriteLine($"Interpolation: {interpolated}");

// String splitting and joining
Console.WriteLine($"\n=== String Splitting and Joining ===");
string sentence = "Apple,Banana,Cherry,Date";
string[] words = sentence.Split(',');
Console.WriteLine("Split result:");
foreach (string word in words)
{
    Console.WriteLine($"- {word}");
}

string rejoined = string.Join(" | ", words);
Console.WriteLine($"Rejoined: {rejoined}");

// String trimming and padding
Console.WriteLine($"\n=== String Trimming and Padding ===");
string messyString = "  Hello World  ";
Console.WriteLine($"Original: '{messyString}'");
Console.WriteLine($"Trimmed: '{messyString.Trim()}'");
Console.WriteLine($"Padded left: '{messyString.Trim().PadLeft(20, '*')}'");
Console.WriteLine($"Padded right: '{messyString.Trim().PadRight(20, '*')}'");

// String replacement
Console.WriteLine($"\n=== String Replacement ===");
string original = "The quick brown fox jumps over the lazy dog";
string replaced = original.Replace("fox", "cat").Replace("dog", "mouse");
Console.WriteLine($"Original: {original}");
Console.WriteLine($"Replaced: {replaced}");

// StringBuilder for efficient string building
Console.WriteLine($"\n=== StringBuilder Example ===");
StringBuilder sb = new StringBuilder();
sb.AppendLine("Line 1");
sb.AppendLine("Line 2");
sb.AppendLine("Line 3");
Console.WriteLine("StringBuilder result:");
Console.WriteLine(sb.ToString());
````

## 8. Array Operations

Arrays are fundamental data structures in C#. Let's explore various array operations.

````python
// Array Operations Examples
Console.WriteLine("=== Array Operations ===");

// Array initialization methods
Console.WriteLine("Array Initialization:");

// Method 1: Initialize with values
int[] numbers1 = { 1, 2, 3, 4, 5 };
Console.WriteLine($"Method 1: [{string.Join(", ", numbers1)}]");

// Method 2: Create and assign
int[] numbers2 = new int[5];
for (int i = 0; i < numbers2.Length; i++)
{
    numbers2[i] = (i + 1) * 10;
}
Console.WriteLine($"Method 2: [{string.Join(", ", numbers2)}]");

// Method 3: Initialize with new keyword
int[] numbers3 = new int[] { 100, 200, 300, 400, 500 };
Console.WriteLine($"Method 3: [{string.Join(", ", numbers3)}]");

// Array properties and basic operations
Console.WriteLine($"\n=== Array Properties ===");
Console.WriteLine($"Length of numbers1: {numbers1.Length}");
Console.WriteLine($"First element: {numbers1[0]}");
Console.WriteLine($"Last element: {numbers1[numbers1.Length - 1]}");

// Array manipulation
Console.WriteLine($"\n=== Array Manipulation ===");
string[] fruits = { "Apple", "Banana", "Cherry", "Date", "Elderberry" };
Console.WriteLine($"Original fruits: [{string.Join(", ", fruits)}]");

// Sorting
Array.Sort(fruits);
Console.WriteLine($"Sorted fruits: [{string.Join(", ", fruits)}]");

// Reversing
Array.Reverse(fruits);
Console.WriteLine($"Reversed fruits: [{string.Join(", ", fruits)}]");

// Finding elements
Console.WriteLine($"\n=== Finding Elements ===");
int[] searchNumbers = { 10, 25, 30, 45, 50, 65, 70, 85, 90 };
Console.WriteLine($"Numbers: [{string.Join(", ", searchNumbers)}]");

int searchValue = 45;
int index = Array.IndexOf(searchNumbers, searchValue);
Console.WriteLine($"Index of {searchValue}: {index}");

bool contains = Array.Exists(searchNumbers, x => x > 80);
Console.WriteLine($"Contains number > 80: {contains}");

// Multi-dimensional arrays
Console.WriteLine($"\n=== Multi-dimensional Arrays ===");

// 2D array
int[,] matrix = {
    { 1, 2, 3 },
    { 4, 5, 6 },
    { 7, 8, 9 }
};

Console.WriteLine("2D Matrix:");
for (int i = 0; i < matrix.GetLength(0); i++)
{
    for (int j = 0; j < matrix.GetLength(1); j++)
    {
        Console.Write($"{matrix[i, j]}\t");
    }
    Console.WriteLine();
}

// Jagged array
Console.WriteLine($"\n=== Jagged Arrays ===");
int[][] jaggedArray = new int[3][];
jaggedArray[0] = new int[] { 1, 2 };
jaggedArray[1] = new int[] { 3, 4, 5, 6 };
jaggedArray[2] = new int[] { 7, 8, 9 };

Console.WriteLine("Jagged Array:");
for (int i = 0; i < jaggedArray.Length; i++)
{
    Console.WriteLine($"Row {i}: [{string.Join(", ", jaggedArray[i])}]");
}

// Array methods
Console.WriteLine($"\n=== Array Methods ===");
int[] workingArray = { 5, 2, 8, 1, 9, 3 };
Console.WriteLine($"Original: [{string.Join(", ", workingArray)}]");

// Copy array
int[] copiedArray = new int[workingArray.Length];
Array.Copy(workingArray, copiedArray, workingArray.Length);
Console.WriteLine($"Copied: [{string.Join(", ", copiedArray)}]");

// Clear part of array
Array.Clear(copiedArray, 2, 2); // Clear 2 elements starting at index 2
Console.WriteLine($"After Clear: [{string.Join(", ", copiedArray)}]");

// Resize array (creates new array)
Array.Resize(ref copiedArray, 8);
Console.WriteLine($"After Resize: [{string.Join(", ", copiedArray)}]");
````

## 9. Dictionary Operations

Dictionaries provide key-value pair storage and are extremely useful for data organization.

````python
// Dictionary Operations Examples
Console.WriteLine("=== Dictionary Operations ===");

// Creating and initializing dictionaries
Dictionary<string, int> ages = new Dictionary<string, int>
{
    { "Alice", 25 },
    { "Bob", 30 },
    { "Charlie", 35 }
};

Console.WriteLine("Initial dictionary:");
foreach (var kvp in ages)
{
    Console.WriteLine($"{kvp.Key}: {kvp.Value}");
}

// Adding and updating elements
Console.WriteLine("\n=== Adding and Updating ===");
ages["David"] = 28; // Add new
ages["Alice"] = 26; // Update existing

Console.WriteLine("After adding David and updating Alice:");
foreach (var kvp in ages)
{
    Console.WriteLine($"{kvp.Key}: {kvp.Value}");
}

// Safe access using TryGetValue
Console.WriteLine("\n=== Safe Access with TryGetValue ===");
string searchKey = "Eve";
if (ages.TryGetValue(searchKey, out int age))
{
    Console.WriteLine($"{searchKey} is {age} years old");
}
else
{
    Console.WriteLine($"{searchKey} not found in dictionary");
}

// Checking for keys and values
Console.WriteLine("\n=== Checking Keys and Values ===");
Console.WriteLine($"Contains key 'Bob': {ages.ContainsKey("Bob")}");
Console.WriteLine($"Contains value 30: {ages.ContainsValue(30)}");

// Dictionary properties
Console.WriteLine($"\nDictionary Count: {ages.Count}");
Console.WriteLine($"Keys: [{string.Join(", ", ages.Keys)}]");
Console.WriteLine($"Values: [{string.Join(", ", ages.Values)}]");

// Removing elements
Console.WriteLine("\n=== Removing Elements ===");
if (ages.Remove("Charlie"))
{
    Console.WriteLine("Charlie removed successfully");
}

Console.WriteLine("After removing Charlie:");
foreach (var kvp in ages)
{
    Console.WriteLine($"{kvp.Key}: {kvp.Value}");
}

// Working with complex data types
Console.WriteLine("\n=== Dictionary with Complex Types ===");

// Dictionary with string key and List<string> value
Dictionary<string, List<string>> studentGrades = new Dictionary<string, List<string>>
{
    { "Math", new List<string> { "A", "B+", "A-" } },
    { "Science", new List<string> { "B", "A", "B+" } },
    { "English", new List<string> { "A-", "A", "B" } }
};

Console.WriteLine("Student grades by subject:");
foreach (var subject in studentGrades)
{
    Console.WriteLine($"{subject.Key}: [{string.Join(", ", subject.Value)}]");
}

// Dictionary methods and LINQ
Console.WriteLine("\n=== Dictionary with LINQ ===");
var highAges = ages.Where(kvp => kvp.Value >= 30).ToDictionary(kvp => kvp.Key, kvp => kvp.Value);
Console.WriteLine("People aged 30 or older:");
foreach (var kvp in highAges)
{
    Console.WriteLine($"{kvp.Key}: {kvp.Value}");
}

// Creating lookup dictionary
Console.WriteLine("\n=== Creating Lookup Dictionary ===");
var people = new[]
{
    new { Name = "Alice", Department = "Engineering" },
    new { Name = "Bob", Department = "Marketing" },
    new { Name = "Charlie", Department = "Engineering" },
    new { Name = "David", Department = "Sales" }
};

var departmentLookup = people.GroupBy(p => p.Department)
                           .ToDictionary(g => g.Key, g => g.Select(p => p.Name).ToList());

Console.WriteLine("People by department:");
foreach (var dept in departmentLookup)
{
    Console.WriteLine($"{dept.Key}: [{string.Join(", ", dept.Value)}]");
}
````

## 10. LINQ Introduction and Examples

LINQ (Language Integrated Query) provides powerful querying capabilities for collections.

````python
// LINQ Examples - Employee and Department Data
Console.WriteLine("=== LINQ Introduction and Examples ===");

// Define Employee and Department classes (from your project)
public class Employee
{
    public int Id { get; set; }
    public string First_name { get; set; }
    public string Last_name { get; set; }
    public decimal Annual_salary { get; set; }
    public bool IsManager { get; set; }
    public int Department { get; set; }
}

public class Department
{
    public int Id { get; set; }
    public string Short_name { get; set; }
    public string Long_name { get; set; }
}

// Create sample data
List<Employee> employees = new List<Employee>
{
    new Employee { Id = 1, First_name = "John", Last_name = "Doe", Annual_salary = 60000m, IsManager = true, Department = 1 },
    new Employee { Id = 2, First_name = "Jane", Last_name = "Smith", Annual_salary = 65000m, IsManager = true, Department = 2 },
    new Employee { Id = 3, First_name = "Bob", Last_name = "Johnson", Annual_salary = 50000m, IsManager = false, Department = 3 },
    new Employee { Id = 4, First_name = "Alice", Last_name = "Brown", Annual_salary = 55000m, IsManager = false, Department = 3 },
    new Employee { Id = 5, First_name = "Charlie", Last_name = "Wilson", Annual_salary = 70000m, IsManager = true, Department = 1 }
};

List<Department> departments = new List<Department>
{
    new Department { Id = 1, Short_name = "HR", Long_name = "Human Resources" },
    new Department { Id = 2, Short_name = "FN", Long_name = "Finance" },
    new Department { Id = 3, Short_name = "TE", Long_name = "Technology" }
};

Console.WriteLine("Sample Data Created:");
Console.WriteLine($"Employees: {employees.Count}");
Console.WriteLine($"Departments: {departments.Count}");

// Basic LINQ operations
Console.WriteLine("\n=== Basic LINQ Operations ===");

// Where - Filtering
var managers = employees.Where(e => e.IsManager).ToList();
Console.WriteLine($"\nManagers ({managers.Count}):");
foreach (var manager in managers)
{
    Console.WriteLine($"- {manager.First_name} {manager.Last_name} (${manager.Annual_salary})");
}

// Select - Projection
var employeeNames = employees.Select(e => $"{e.First_name} {e.Last_name}").ToList();
Console.WriteLine($"\nAll Employee Names:");
foreach (var name in employeeNames)
{
    Console.WriteLine($"- {name}");
}

// OrderBy - Sorting
var employeesBySalary = employees.OrderByDescending(e => e.Annual_salary).ToList();
Console.WriteLine($"\nEmployees by Salary (Highest to Lowest):");
foreach (var emp in employeesBySalary)
{
    Console.WriteLine($"- {emp.First_name} {emp.Last_name}: ${emp.Annual_salary}");
}

// GroupBy - Grouping
var employeesByDept = employees.GroupBy(e => e.Department).ToList();
Console.WriteLine($"\nEmployees Grouped by Department:");
foreach (var group in employeesByDept)
{
    var deptName = departments.First(d => d.Id == group.Key).Long_name;
    Console.WriteLine($"  {deptName} ({group.Count()} employees):");
    foreach (var emp in group)
    {
        Console.WriteLine($"    - {emp.First_name} {emp.Last_name}");
    }
}

// Aggregate functions
Console.WriteLine("\n=== LINQ Aggregate Functions ===");
Console.WriteLine($"Total employees: {employees.Count()}");
Console.WriteLine($"Average salary: ${employees.Average(e => e.Annual_salary):F2}");
Console.WriteLine($"Highest salary: ${employees.Max(e => e.Annual_salary)}");
Console.WriteLine($"Lowest salary: ${employees.Min(e => e.Annual_salary)}");
Console.WriteLine($"Total salary budget: ${employees.Sum(e => e.Annual_salary)}");

// Join operations
Console.WriteLine("\n=== LINQ Join Operations ===");
var employeeWithDept = from emp in employees
                       join dept in departments on emp.Department equals dept.Id
                       select new 
                       {
                           Name = $"{emp.First_name} {emp.Last_name}",
                           Salary = emp.Annual_salary,
                           Department = dept.Long_name,
                           IsManager = emp.IsManager
                       };

Console.WriteLine("Employee Details with Department:");
foreach (var item in employeeWithDept)
{
    Console.WriteLine($"- {item.Name} | {item.Department} | ${item.Salary} | Manager: {item.IsManager}");
}

// Complex LINQ queries
Console.WriteLine("\n=== Complex LINQ Queries ===");

// Find departments with average salary > 60000
var highPayingDepts = from emp in employees
                      join dept in departments on emp.Department equals dept.Id
                      group emp by new { dept.Id, dept.Long_name } into g
                      where g.Average(e => e.Annual_salary) > 60000
                      select new
                      {
                          Department = g.Key.Long_name,
                          AvgSalary = g.Average(e => e.Annual_salary),
                          EmployeeCount = g.Count()
                      };

Console.WriteLine("High-paying departments (avg salary > $60,000):");
foreach (var dept in highPayingDepts)
{
    Console.WriteLine($"- {dept.Department}: Avg ${dept.AvgSalary:F2} ({dept.EmployeeCount} employees)");
}

// Any and All methods
Console.WriteLine("\n=== LINQ Any/All Methods ===");
Console.WriteLine($"Any managers earning > $65,000: {employees.Any(e => e.IsManager && e.Annual_salary > 65000)}");
Console.WriteLine($"All employees earn > $40,000: {employees.All(e => e.Annual_salary > 40000)}");
Console.WriteLine($"Technology dept has managers: {employees.Any(e => e.Department == 3 && e.IsManager)}");
````

## 11. Exception Handling

Exception handling is crucial for building robust applications. Let's explore try-catch-finally blocks.

````python
// Exception Handling Examples
Console.WriteLine("=== Exception Handling Examples ===");

// Basic try-catch
Console.WriteLine("1. Basic try-catch:");
try
{
    int number = 10;
    int zero = 0;
    int result = number / zero; // This will throw DivideByZeroException
    Console.WriteLine($"Result: {result}");
}
catch (DivideByZeroException ex)
{
    Console.WriteLine($"❌ Division by zero error: {ex.Message}");
}

// Multiple catch blocks
Console.WriteLine("\n2. Multiple catch blocks:");
try
{
    string[] array = { "1", "2", "abc" };
    for (int i = 0; i <= 3; i++) // Will go out of bounds
    {
        int value = Convert.ToInt32(array[i]);
        Console.WriteLine($"Converted: {value}");
    }
}
catch (IndexOutOfRangeException ex)
{
    Console.WriteLine($"❌ Array index error: {ex.Message}");
}
catch (FormatException ex)
{
    Console.WriteLine($"❌ Format error: {ex.Message}");
}
catch (Exception ex)
{
    Console.WriteLine($"❌ General error: {ex.Message}");
}

// Try-catch-finally
Console.WriteLine("\n3. Try-catch-finally:");
try
{
    Console.WriteLine("  Executing code in try block");
    throw new InvalidOperationException("Custom exception for demonstration");
}
catch (InvalidOperationException ex)
{
    Console.WriteLine($"  ❌ Caught exception: {ex.Message}");
}
finally
{
    Console.WriteLine("  ✅ Finally block always executes");
}

// Custom exception handling
Console.WriteLine("\n4. Custom exception handling:");

void ValidateAge(int age)
{
    if (age < 0)
        throw new ArgumentException("Age cannot be negative");
    if (age > 150)
        throw new ArgumentException("Age cannot be greater than 150");
    Console.WriteLine($"  ✅ Valid age: {age}");
}

try
{
    ValidateAge(25);
    ValidateAge(-5);  // This will throw
}
catch (ArgumentException ex)
{
    Console.WriteLine($"  ❌ Validation error: {ex.Message}");
}

// Exception with data
Console.WriteLine("\n5. Exception with additional data:");
try
{
    var ex = new Exception("Something went wrong");
    ex.Data.Add("ErrorCode", "E001");
    ex.Data.Add("Timestamp", DateTime.Now);
    throw ex;
}
catch (Exception ex)
{
    Console.WriteLine($"  ❌ Error: {ex.Message}");
    foreach (var key in ex.Data.Keys)
    {
        Console.WriteLine($"    {key}: {ex.Data[key]}");
    }
}

// Nested try-catch
Console.WriteLine("\n6. Nested try-catch:");
try
{
    Console.WriteLine("  Outer try block");
    try
    {
        Console.WriteLine("    Inner try block");
        throw new InvalidCastException("Inner exception");
    }
    catch (InvalidCastException innerEx)
    {
        Console.WriteLine($"    ❌ Inner catch: {innerEx.Message}");
        throw new Exception("Outer exception", innerEx); // Re-throw with wrapper
    }
}
catch (Exception outerEx)
{
    Console.WriteLine($"  ❌ Outer catch: {outerEx.Message}");
    if (outerEx.InnerException != null)
    {
        Console.WriteLine($"    Inner exception: {outerEx.InnerException.Message}");
    }
}

// Using statement for automatic resource disposal
Console.WriteLine("\n7. Using statement for resource management:");
try
{
    using (var stringWriter = new StringWriter())
    {
        stringWriter.WriteLine("This is a test");
        Console.WriteLine($"  ✅ Content: {stringWriter.ToString()}");
        // stringWriter will be automatically disposed
    }
    Console.WriteLine("  ✅ StringWriter disposed automatically");
}
catch (Exception ex)
{
    Console.WriteLine($"  ❌ Error with using statement: {ex.Message}");
}

// Exception propagation example
Console.WriteLine("\n8. Exception propagation:");

void MethodThatThrows()
{
    throw new NotSupportedException("This method is not supported");
}

void MethodThatCalls()
{
    MethodThatThrows(); // Exception will propagate up
}

try
{
    MethodThatCalls();
}
catch (NotSupportedException ex)
{
    Console.WriteLine($"  ❌ Caught propagated exception: {ex.Message}");
    Console.WriteLine($"  Stack trace snippet: {ex.StackTrace?.Split('\n')[0]}");
}
````

## 12. Exercise: FizzBuzz Implementation

The classic FizzBuzz problem: print numbers 1 to N, but replace multiples of 3 with "Fizz", multiples of 5 with "Buzz", and multiples of both with "FizzBuzz".

````python
// FizzBuzz Exercise Implementation
Console.WriteLine("=== FizzBuzz Exercise ===");

// Basic FizzBuzz implementation
void FizzBuzz(int number)
{
    Console.WriteLine($"\nFizzBuzz from 1 to {number}:");
    var sb = new StringBuilder();
    
    for (int i = 1; i <= number; i++)
    {
        bool divisibleBy3 = i % 3 == 0;
        bool divisibleBy5 = i % 5 == 0;

        if (divisibleBy3 && divisibleBy5)
            sb.AppendLine($"{i}: FizzBuzz");
        else if (divisibleBy3)
            sb.AppendLine($"{i}: Fizz");
        else if (divisibleBy5)
            sb.AppendLine($"{i}: Buzz");
        else
            sb.AppendLine($"{i}: {i}");
    }
    
    Console.Write(sb.ToString());
}

// Test with different numbers
Console.WriteLine("Testing FizzBuzz with 15:");
FizzBuzz(15);

Console.WriteLine("\nTesting FizzBuzz with 30:");
FizzBuzz(30);

// Alternative implementation using switch expression
Console.WriteLine("\n=== Alternative FizzBuzz using Switch Expression ===");

string FizzBuzzSwitch(int number) => (number % 3, number % 5) switch
{
    (0, 0) => "FizzBuzz",
    (0, _) => "Fizz",
    (_, 0) => "Buzz",
    _ => number.ToString()
};

Console.WriteLine("FizzBuzz 1-20 using switch expression:");
for (int i = 1; i <= 20; i++)
{
    Console.WriteLine($"{i}: {FizzBuzzSwitch(i)}");
}

// Extended FizzBuzz with custom rules
Console.WriteLine("\n=== Extended FizzBuzz with Custom Rules ===");

void ExtendedFizzBuzz(int number, Dictionary<int, string> rules)
{
    Console.WriteLine($"\nExtended FizzBuzz from 1 to {number} with custom rules:");
    
    for (int i = 1; i <= number; i++)
    {
        var result = new StringBuilder();
        
        foreach (var rule in rules.OrderBy(r => r.Key))
        {
            if (i % rule.Key == 0)
                result.Append(rule.Value);
        }
        
        Console.WriteLine($"{i}: {(result.Length > 0 ? result.ToString() : i.ToString())}");
    }
}

// Test extended FizzBuzz
var customRules = new Dictionary<int, string>
{
    { 3, "Fizz" },
    { 5, "Buzz" },
    { 7, "Bang" }
};

ExtendedFizzBuzz(21, customRules);

// Performance comparison
Console.WriteLine("\n=== Performance Comparison ===");

var stopwatch = System.Diagnostics.Stopwatch.StartNew();

// StringBuilder version
var sb1 = new StringBuilder();
for (int i = 1; i <= 1000; i++)
{
    bool divisibleBy3 = i % 3 == 0;
    bool divisibleBy5 = i % 5 == 0;

    if (divisibleBy3 && divisibleBy5)
        sb1.Append("FizzBuzz,");
    else if (divisibleBy3)
        sb1.Append("Fizz,");
    else if (divisibleBy5)
        sb1.Append("Buzz,");
    else
        sb1.Append($"{i},");
}

stopwatch.Stop();
Console.WriteLine($"StringBuilder version (1000 iterations): {stopwatch.ElapsedMilliseconds} ms");

// LINQ version
stopwatch.Restart();
var linqResult = Enumerable.Range(1, 1000)
    .Select(i => (i % 3 == 0 && i % 5 == 0) ? "FizzBuzz" :
                 (i % 3 == 0) ? "Fizz" :
                 (i % 5 == 0) ? "Buzz" : i.ToString())
    .ToArray();
stopwatch.Stop();
Console.WriteLine($"LINQ version (1000 iterations): {stopwatch.ElapsedMilliseconds} ms");
````

## 13. Exercise: Times Table Generator

Create multiplication tables for any given number with various formatting options.

````python
// Times Table Exercise Implementation
Console.WriteLine("=== Times Table Generator ===");

// Basic times table implementation
void GenerateTimesTable(int number, int maxMultiplier = 10)
{
    Console.WriteLine($"\nTimes table for {number}:");
    Console.WriteLine(new string('-', 20));
    
    for (int i = 1; i <= maxMultiplier; i++)
    {
        Console.WriteLine($"{i,2} × {number,2} = {i * number,3}");
    }
}

// Test with different numbers
GenerateTimesTable(7);
GenerateTimesTable(12, 12);

// Enhanced times table with formatting
Console.WriteLine("\n=== Enhanced Times Table with Better Formatting ===");

void GenerateEnhancedTimesTable(int number, int maxMultiplier = 10)
{
    Console.WriteLine($"\n╔══════════════════════════════╗");
    Console.WriteLine($"║     Times Table for {number,2}       ║");
    Console.WriteLine($"╠══════════════════════════════╣");
    
    for (int i = 1; i <= maxMultiplier; i++)
    {
        int result = i * number;
        Console.WriteLine($"║ {i,2} × {number,2} = {result,3}                ║");
    }
    
    Console.WriteLine($"╚══════════════════════════════╝");
}

GenerateEnhancedTimesTable(9);

// Multiple tables side by side
Console.WriteLine("\n=== Multiple Times Tables Side by Side ===");

void GenerateMultipleTables(int[] numbers, int maxMultiplier = 10)
{
    // Header
    Console.Write("Row ");
    foreach (int num in numbers)
    {
        Console.Write($"│ {num,2}x │");
    }
    Console.WriteLine();
    
    Console.Write("────");
    foreach (int num in numbers)
    {
        Console.Write("┼─────┼");
    }
    Console.WriteLine();
    
    // Table rows
    for (int i = 1; i <= maxMultiplier; i++)
    {
        Console.Write($"{i,2}  ");
        foreach (int num in numbers)
        {
            Console.Write($"│{i * num,4} │");
        }
        Console.WriteLine();
    }
}

GenerateMultipleTables(new int[] { 2, 3, 4, 5 }, 12);

// Interactive times table quiz
Console.WriteLine("\n=== Times Table Quiz Generator ===");

void TimesTableQuiz(int number, int numberOfQuestions = 5)
{
    Console.WriteLine($"\nTimes Table Quiz for {number} (Answer format: just the number)");
    Console.WriteLine(new string('=', 40));
    
    Random random = new Random();
    int correctAnswers = 0;
    
    for (int q = 1; q <= numberOfQuestions; q++)
    {
        int multiplier = random.Next(1, 13); // Random number 1-12
        int correctAnswer = number * multiplier;
        
        Console.WriteLine($"\nQuestion {q}: {number} × {multiplier} = ?");
        
        // For demonstration, we'll show both question and answer
        Console.WriteLine($"Answer: {correctAnswer}");
        correctAnswers++; // Assume all correct for demo
    }
    
    double percentage = (double)correctAnswers / numberOfQuestions * 100;
    Console.WriteLine($"\nQuiz Complete! Score: {correctAnswers}/{numberOfQuestions} ({percentage:F1}%)");
    
    string grade = percentage switch
    {
        >= 90 => "Excellent! 🌟",
        >= 80 => "Very Good! 👍",
        >= 70 => "Good! 👌",
        >= 60 => "Fair 📚",
        _ => "Keep practicing! 💪"
    };
    
    Console.WriteLine($"Grade: {grade}");
}

TimesTableQuiz(6, 3);

// Times table patterns
Console.WriteLine("\n=== Times Table Patterns ===");

void ShowTimesTablePatterns(int number)
{
    Console.WriteLine($"\nPatterns in the {number} times table:");
    
    // Digital root pattern
    Console.WriteLine("\nDigital roots (sum of digits until single digit):");
    for (int i = 1; i <= 10; i++)
    {
        int result = number * i;
        int digitalRoot = GetDigitalRoot(result);
        Console.WriteLine($"{number} × {i} = {result} → Digital root: {digitalRoot}");
    }
}

int GetDigitalRoot(int number)
{
    while (number >= 10)
    {
        int sum = 0;
        while (number > 0)
        {
            sum += number % 10;
            number /= 10;
        }
        number = sum;
    }
    return number;
}

ShowTimesTablePatterns(9);

// Generate times table as 2D array
Console.WriteLine("\n=== Times Table as 2D Array ===");

int[,] GenerateTimesTableArray(int maxNumber)
{
    int[,] table = new int[maxNumber + 1, maxNumber + 1];
    
    for (int i = 1; i <= maxNumber; i++)
    {
        for (int j = 1; j <= maxNumber; j++)
        {
            table[i, j] = i * j;
        }
    }
    
    return table;
}

void DisplayTimesTableArray(int[,] table, int size)
{
    Console.WriteLine($"\nComplete times table (1-{size}):");
    
    // Header row
    Console.Write("   ");
    for (int j = 1; j <= size; j++)
    {
        Console.Write($"{j,4}");
    }
    Console.WriteLine();
    
    // Separator
    Console.Write("   ");
    for (int j = 1; j <= size; j++)
    {
        Console.Write("────");
    }
    Console.WriteLine();
    
    // Data rows
    for (int i = 1; i <= size; i++)
    {
        Console.Write($"{i,2}│");
        for (int j = 1; j <= size; j++)
        {
            Console.Write($"{table[i, j],4}");
        }
        Console.WriteLine();
    }
}

var timesTableArray = GenerateTimesTableArray(6);
DisplayTimesTableArray(timesTableArray, 6);
````

## 14. Exercise: Password Validation

Create a robust password validation system with multiple security requirements.

````python
// Password Validation Exercise Implementation
Console.WriteLine("=== Password Validation System ===");

// Simple password validation (from your original project)
void SimplePasswordValidation()
{
    Console.WriteLine("Simple Password Validation (Password: 123):");
    
    int correctPassword = 123;
    int attempts = 3;
    
    // Simulate different password attempts
    int[] testPasswords = { 456, 789, 123 };
    
    foreach (int password in testPasswords)
    {
        if (attempts <= 0) break;
        
        Console.WriteLine($"\nAttempting password: {password}");
        
        if (password == correctPassword)
        {
            Console.WriteLine("✅ Correct password! Access granted.");
            break;
        }
        
        attempts--;
        
        if (attempts > 0)
        {
            Console.WriteLine($"❌ Incorrect. You have {attempts} attempt(s) left.");
        }
        else
        {
            Console.WriteLine("❌ No attempts left. Access denied.");
        }
    }
}

SimplePasswordValidation();

// Advanced password validation with multiple criteria
Console.WriteLine("\n=== Advanced Password Validation ===");

public class PasswordValidator
{
    public static bool ValidatePassword(string password, out List<string> errors)
    {
        errors = new List<string>();
        
        if (string.IsNullOrEmpty(password))
        {
            errors.Add("Password cannot be empty");
            return false;
        }
        
        if (password.Length < 8)
            errors.Add("Password must be at least 8 characters long");
            
        if (password.Length > 128)
            errors.Add("Password must be less than 128 characters long");
            
        if (!password.Any(char.IsUpper))
            errors.Add("Password must contain at least one uppercase letter");
            
        if (!password.Any(char.IsLower))
            errors.Add("Password must contain at least one lowercase letter");
            
        if (!password.Any(char.IsDigit))
            errors.Add("Password must contain at least one digit");
            
        if (!password.Any(ch => "!@#$%^&*()_+-=[]{}|;:,.<>?".Contains(ch)))
            errors.Add("Password must contain at least one special character");
            
        if (HasCommonPatterns(password))
            errors.Add("Password contains common patterns and is not secure");
            
        return errors.Count == 0;
    }
    
    private static bool HasCommonPatterns(string password)
    {
        string[] commonPatterns = { "123", "abc", "password", "admin", "qwerty" };
        return commonPatterns.Any(pattern => 
            password.ToLower().Contains(pattern.ToLower()));
    }
    
    public static int CalculatePasswordStrength(string password)
    {
        int strength = 0;
        
        if (password.Length >= 8) strength += 1;
        if (password.Length >= 12) strength += 1;
        if (password.Any(char.IsUpper)) strength += 1;
        if (password.Any(char.IsLower)) strength += 1;
        if (password.Any(char.IsDigit)) strength += 1;
        if (password.Any(ch => "!@#$%^&*()_+-=[]{}|;:,.<>?".Contains(ch))) strength += 1;
        if (password.Length >= 16) strength += 1;
        
        return Math.Min(strength, 5); // Max strength of 5
    }
}

// Test different passwords
string[] testPasswords = {
    "weak",
    "password123",
    "MyPassword123",
    "MyP@ssw0rd123",
    "C0mpl3x!P@ssw0rd#2024",
    ""
};

Console.WriteLine("Testing password validation:");
Console.WriteLine(new string('=', 50));

foreach (string pwd in testPasswords)
{
    Console.WriteLine($"\nTesting password: '{pwd}'");
    
    bool isValid = PasswordValidator.ValidatePassword(pwd, out List<string> errors);
    int strength = PasswordValidator.CalculatePasswordStrength(pwd);
    
    if (isValid)
    {
        Console.WriteLine("✅ Password is valid!");
    }
    else
    {
        Console.WriteLine("❌ Password validation failed:");
        foreach (string error in errors)
        {
            Console.WriteLine($"   - {error}");
        }
    }
    
    string strengthText = strength switch
    {
        0 => "Very Weak 🔴",
        1 => "Weak 🟠", 
        2 => "Fair 🟡",
        3 => "Good 🟢",
        4 => "Strong 🔵",
        5 => "Very Strong 🟣",
        _ => "Unknown"
    };
    
    Console.WriteLine($"   Strength: {strengthText} ({strength}/5)");
}

// Password generator
Console.WriteLine("\n=== Password Generator ===");

public class PasswordGenerator
{
    private static readonly string UppercaseChars = "ABCDEFGHIJKLMNOPQRSTUVWXYZ";
    private static readonly string LowercaseChars = "abcdefghijklmnopqrstuvwxyz";
    private static readonly string DigitChars = "0123456789";
    private static readonly string SpecialChars = "!@#$%^&*()_+-=[]{}|;:,.<>?";
    
    public static string GeneratePassword(int length = 12, bool includeSpecial = true)
    {
        var random = new Random();
        var allChars = UppercaseChars + LowercaseChars + DigitChars;
        if (includeSpecial) allChars += SpecialChars;
        
        var password = new StringBuilder();
        
        // Ensure at least one character from each required category
        password.Append(UppercaseChars[random.Next(UppercaseChars.Length)]);
        password.Append(LowercaseChars[random.Next(LowercaseChars.Length)]);
        password.Append(DigitChars[random.Next(DigitChars.Length)]);
        if (includeSpecial)
            password.Append(SpecialChars[random.Next(SpecialChars.Length)]);
        
        // Fill the rest randomly
        int remainingLength = length - password.Length;
        for (int i = 0; i < remainingLength; i++)
        {
            password.Append(allChars[random.Next(allChars.Length)]);
        }
        
        // Shuffle the password
        return new string(password.ToString().OrderBy(x => random.Next()).ToArray());
    }
}

Console.WriteLine("Generated secure passwords:");
for (int i = 0; i < 5; i++)
{
    string generatedPassword = PasswordGenerator.GeneratePassword(12, true);
    int strength = PasswordValidator.CalculatePasswordStrength(generatedPassword);
    bool isValid = PasswordValidator.ValidatePassword(generatedPassword, out _);
    
    Console.WriteLine($"{i + 1}. {generatedPassword} | Strength: {strength}/5 | Valid: {(isValid ? "✅" : "❌")}");
}

// Password security tips
Console.WriteLine("\n=== Password Security Tips ===");
Console.WriteLine("✅ Use at least 12 characters");
Console.WriteLine("✅ Mix uppercase and lowercase letters");
Console.WriteLine("✅ Include numbers and special characters");
Console.WriteLine("✅ Avoid common words and patterns");
Console.WriteLine("✅ Use unique passwords for each account");
Console.WriteLine("✅ Consider using a password manager");
Console.WriteLine("✅ Enable two-factor authentication when available");
````
