````polyglot-notebook
#!csharp
using System;
using System.Linq;
using System.Collections.Generic;
using System.Text.Json;

// Data (same as Program.cs)
var employees = new List<Dictionary<string, object>>
{
    new() { ["Id"] = 1, ["FirstName"] = "Bob", ["LastName"] = "Jones", ["AnnualSalary"] = 60000.3m, ["IsManager"] = true, ["DepartmentId"] = 1 },
    new() { ["Id"] = 2, ["FirstName"] = "Sarah", ["LastName"] = "Jameson", ["AnnualSalary"] = 80000.1m, ["IsManager"] = true, ["DepartmentId"] = 2 },
    new() { ["Id"] = 3, ["FirstName"] = "Douglas", ["LastName"] = "Roberts", ["AnnualSalary"] = 40000.2m, ["IsManager"] = false, ["DepartmentId"] = 2 },
    new() { ["Id"] = 4, ["FirstName"] = "Jane", ["LastName"] = "Stevens", ["AnnualSalary"] = 30000.2m, ["IsManager"] = false, ["DepartmentId"] = 3 },
    new() { ["Id"] = 5, ["FirstName"] = "Kostantin", ["LastName"] = "Spenst", ["AnnualSalary"] = 300000.2m, ["IsManager"] = false, ["DepartmentId"] = 3 }
};

var departments = new List<Dictionary<string, object>>
{
    new() { ["Id"] = 1, ["ShortName"] = "HR", ["LongName"] = "Human Resources" },
    new() { ["Id"] = 2, ["ShortName"] = "FN", ["LongName"] = "Finance" },
    new() { ["Id"] = 3, ["ShortName"] = "TE", ["LongName"] = "Technology" }
};

// Example: Where (high earners)
var highEarners = employees.Where(e => (decimal)e["AnnualSalary"] > 50000).ToList();
Console.WriteLine("High Earners JSON:");
Console.WriteLine(JsonSerializer.Serialize(highEarners, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: Select (names)
var names = employees.Select(e => $"{e["FirstName"]} {e["LastName"]}").ToList();
Console.WriteLine("Names JSON:");
Console.WriteLine(JsonSerializer.Serialize(names, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: OrderBy
var ordered = employees.OrderBy(e => (string)e["LastName"]).ThenBy(e => (int)e["Id"]).Select(e => new { LastName = e["LastName"], FirstName = e["FirstName"] }).ToList();
Console.WriteLine("Ordered JSON:");
Console.WriteLine(JsonSerializer.Serialize(ordered, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: ToList (managers)
var managers = employees.Where(e => (bool)e["IsManager"]).ToList();
Console.WriteLine("Managers JSON:");
Console.WriteLine(JsonSerializer.Serialize(managers, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: Chaining (first names)
var result = employees.Where(e => (decimal)e["AnnualSalary"] > 50000).OrderBy(e => (string)e["LastName"]).Select(e => e["FirstName"]).ToList();
Console.WriteLine("Chained FirstNames JSON:");
Console.WriteLine(JsonSerializer.Serialize(result, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: SelectMany (characters)
var allChars = employees.SelectMany(e => ((string)e["FirstName"]).ToCharArray()).ToList();
Console.WriteLine("AllChars JSON:");
Console.WriteLine(JsonSerializer.Serialize(allChars, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: Zip (firstName - department shortName)
var zipped = employees.Zip(departments, (e, d) => $"{e["FirstName"]} - {d["ShortName"]}").ToList();
Console.WriteLine("Zipped JSON:");
Console.WriteLine(JsonSerializer.Serialize(zipped, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: Join (inner join)
var joined = employees.Join(
    departments,
    e => e["DepartmentId"],
    d => d["Id"],
    (e, d) => new { Name = $"{e["FirstName"]} {e["LastName"]}", Department = d["LongName"] }
).ToList();
Console.WriteLine("Joined JSON:");
Console.WriteLine(JsonSerializer.Serialize(joined, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: GroupJoin (departments -> employees hierarchical)
var groupJoin = departments.GroupJoin(
    employees,
    d => d["Id"],
    e => e["DepartmentId"],
    (d, emps) => new { Department = d["LongName"], Employees = emps }
).ToList();
Console.WriteLine("GroupJoin JSON:");
Console.WriteLine(JsonSerializer.Serialize(groupJoin, new JsonSerializerOptions { WriteIndented = true }));
````

````polyglot-notebook
#!csharp
// Example: GroupBy
var grouped = employees.GroupBy(e => e["DepartmentId"]).Select(g => new { DepartmentId = g.Key, Employees = g.Select(e => new { Id = e["Id"], FirstName = e["FirstName"], LastName = e["LastName"] }) }).ToList();
Console.WriteLine("Grouped JSON:");
Console.WriteLine(JsonSerializer.Serialize(grouped, new JsonSerializerOptions { WriteIndented = true }));
````
