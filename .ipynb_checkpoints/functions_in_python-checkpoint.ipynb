# 🧠 What is a Function?

A **function** is a reusable block of code that performs a specific task.

In Python, we define a function using the `def` keyword.

Think of it like a machine: you give it some input, it does something, and (optionally) gives you back a result.

```python
def greet():
    print("Hello, student!")

# Call the function
greet()
```

### ✅ Explanation:
- `def greet():` defines a function named `greet`.
- `print("Hello, student!")` is the task it performs.
- `greet()` is how we call the function.

```python
def greet_by_name(name):
    print(f"Hello, {name}!")

# Try it with your name
greet_by_name("Alice")
```

### 📌 Parameters:
A **parameter** is a variable you pass into a function. Here, `name` is the parameter.

```python
def add(a, b):
    return a + b

# Try different values
result = add(5, 3)
print("The sum is:", result)
```

### 🔁 Return Values:
A function can return a result using the `return` keyword.

```python
# Task: Write a function called `square` that takes a number and returns its square.

def square(x):
    return x ** 2

print(square(4))  # Expected: 16
```

```python
# Task: Write a function that checks if a number is even

def is_even(n):
    return n % 2 == 0

print(is_even(10))  # True
print(is_even(7))   # False
```

# 🎯 Summary

- Use `def` to define a function.
- Use `()` to call a function.
- You can pass parameters (inputs).
- You can return values (outputs).

👉 Try modifying the examples or writing your own functions!
